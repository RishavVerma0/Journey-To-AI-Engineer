## List

```python
revenues = [50, 60, 70, 80, 90]
type(revenues)
```

```python
revenues[1:4]
```

```python

expenses = [10, 20, 30, 40, 50]
```

```python
# for i in revenues:
#     print(i)

for i in revenues:
    print(f"Revenues: {i}")
```

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    gain = profit * 100 / revenues[i]
    print(f"Margin ({i}) {gain}")
```

```python
for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    print(f"Margin Q{i+1}: {margin:.2f}%")
```

```python
margins = []

for i in range(len(revenues)):
    profit = revenues[i] - expenses[i]
    margin = profit * 100 / revenues[i]
    margins.append(margin)

print(margins)
```

## Dictionary

```python
fin = {
    "revenue": 50,
    "expenses" : 10
}
type(fin)
```

```python
fin["profit"] = fin["revenue"] - fin["expenses"]
fin
```

```python
financials = [
    {"revenue": 60, "expense": 20},
    {"revenue": 70, "expense": 30},
    {"revenue": 80, "expense": 40}
]
```

```python
for i in range(len(financials)):
    profit = financials[i]["revenue"] - financials[i]["expense"]
    margin = profit * 100 / financials[i]["revenue"]
    print(f"Margin Q{i+1}: {margin:.2f}%")
```
