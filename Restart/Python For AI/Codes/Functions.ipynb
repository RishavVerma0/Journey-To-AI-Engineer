```python
def calculate_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin

calculate_margin(1000, 800)
```

```python
def calculate_margin(revenue, expenses):
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin, profit

calculate_margin(2000, 800)
```

### Typehint
#### It allows another person working on your code to get an idea about the datatypes

```python
def calculate_margin(revenue: float, expenses: float) -> tuple[float, float]:
    profit = revenue - expenses
    margin = profit * 100 / revenue
    return margin, profit

calculate_margin(2000, 800)
```

### Dynamic Inputs

```python
def calculate_sum(*args):
    return sum(args)

print(calculate_sum(1, 2, 3, 4, 5))
```

```python
def total_expenses(**kargs):
    total = 0
    for key, value in kargs.items():
        total += value
    return total

total_expenses(food=100, rent=500, utilities=200)
```
