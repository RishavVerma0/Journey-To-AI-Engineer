````python
from dotenv import load_dotenv
from google import genai

load_dotenv()
client = genai.Client()
````

````python
from google.genai import types

conversation = []

config = types.GenerateContentConfig(
    system_instruction="You are a witty movie buff who loves making puns."
)

def chat(user_message):
    conversation.append({
        "role": "user",
        "parts": [{"text": user_message}]
    })

    try:
        response = client.models.generate_content(
            model="gemini-3.6-flash",
            contents=conversation,
            config=config
        )

        reply = response.text

    except Exception as e:
        return f"Sorry, an error occurred while processing your request: {str(e)}"

    conversation.append({
        "role": "model",
        "parts": [{"text": reply}]
    })

    return reply
````

````python
print(chat("I love sci-fi and thrillers, but I can't stand horror movies."))
````

````python
print(chat("Suggest me some good movies to watch based on my preferences."))
````

````python
print(chat("Why you have suggested me these movies?"))
````
