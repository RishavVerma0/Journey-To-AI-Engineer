```python
import os
import datetime
import requests
import pytz
import yaml

from dotenv import load_dotenv

from smolagents import (
    CodeAgent,
    DuckDuckGoSearchTool,
    LiteLLMModel,
    load_tool,
    tool
)

from smolagents import FinalAnswerTool
from smolagents import GradioUI

# Load environment variables
load_dotenv()
```

```python
# ----------------------------------
# Custom Tool
# ----------------------------------

@tool
def my_custom_tool(arg1: str, arg2: int) -> str:
    """A custom tool for processing user inputs.

    Args:
        arg1: A string input.
        arg2: An integer input.
    """
    return f"Processed {arg1} with value {arg2}"
```

```python
# ----------------------------------
# Timezone Tool
# ----------------------------------

@tool
def get_current_time_in_timezone(timezone: str) -> str:
    """Fetches the current local time in a specified timezone.

    Args:
        timezone: A valid timezone (e.g., 'Asia/Kolkata').
    """
    try:
        tz = pytz.timezone(timezone)

        local_time = datetime.datetime.now(tz).strftime(
            "%Y-%m-%d %H:%M:%S"
        )

        return f"The current local time in {timezone} is: {local_time}"

    except Exception as e:
        return f"Error fetching time: {str(e)}"
```

```python
import os
from dotenv import load_dotenv
from smolagents import LiteLLMModel

load_dotenv()

groq_api_key = os.getenv("GROQ_API_KEY")
if not groq_api_key:
    raise ValueError("GROQ_API_KEY is missing")

model = LiteLLMModel(
    model_id="groq/openai/gpt-oss-120b",
    api_key=groq_api_key,
    temperature=0.1,
    max_tokens=2048,
)
```

```python
# ----------------------------------
# Final Answer Tool
# ----------------------------------

final_answer = FinalAnswerTool()
```

```python
# ----------------------------------
# Image Generation Tool
# ----------------------------------

image_generation_tool = load_tool(
    "agents-course/text-to-image",
    trust_remote_code=True
)
```

```python
import importlib.resources
import yaml

prompt_templates = yaml.safe_load(
    importlib.resources.files("smolagents.prompts").joinpath("code_agent.yaml").read_text()
)
```

```python
from smolagents import VisitWebpageTool

agent = CodeAgent(
    model=model,
    tools=[
    final_answer,
    get_current_time_in_timezone,
    DuckDuckGoSearchTool(max_results=3),
    VisitWebpageTool(max_output_length=3000),
    image_generation_tool,
    ],
    additional_authorized_imports=["pandas", "numpy", "json", "csv"],
    max_steps=5,
    verbosity_level=1,
    planning_interval=None,
    name=None,
    description=None,
    prompt_templates=prompt_templates,
)
```

```python
# ----------------------------------
# Launch Gradio Interface
# ----------------------------------

GradioUI(agent).launch()
```
