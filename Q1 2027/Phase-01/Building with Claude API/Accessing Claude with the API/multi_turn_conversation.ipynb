```python
# Load Env variables from .env file

from dotenv import load_dotenv
load_dotenv()
import os 
```

```python
# Create API Client

from anthropic import Anthropic 

client = Anthropic(
    base_url = os.getenv("ANTHROPIC_BASE_URL"),
    api_key = os.getenv("ANTHROPIC_AUTH_TOKEN"),
)

model_name = os.getenv("ANTHROPIC_MODEL")
```

```python
def add_user_message(messages, text):
    user_message = {'role': 'user', 'content': text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {'role': 'assistant', 'content': text}
    messages.append(assistant_message)

def chat(messages):
    messages = client.messages.create(
        model=model_name,
        max_tokens=1000,
        messages=messages
    )
    return messages.content[0].text

```

```python
# Make starting list of messages
message = []

# Add in the initial user question
add_user_message(message, "What is Quantum Computing?")

# Pass the list of messages to chat to get the answer
answer = chat(message)

# Take the answer and add it to the list of assistant messages
add_assistant_message(message, answer)

# Add in the user's follow-up question
add_user_message(message, "Can you explain that further?")

# Call chat again with the list of messages to get the final answer
answer = chat(message)
answer
```
