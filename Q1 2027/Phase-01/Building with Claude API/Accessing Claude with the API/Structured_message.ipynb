````python
# load env variables from .env file
from dotenv import load_dotenv
load_dotenv()
import os
````

````python
# create api client 
from anthropic import Anthropic
client = Anthropic(
    base_url = os.getenv("ANTHROPIC_BASE_URL"),
    api_key = os.getenv("ANTHROPIC_AUTH_TOKEN"),
)

model_name = os.getenv("ANTHROPIC_MODEL")

print("Base URL configured:", bool(os.getenv("ANTHROPIC_BASE_URL")))

print("API token configured:", bool(os.getenv("ANTHROPIC_AUTH_TOKEN")))

print("Model configured:", bool(model_name))
````

````python
def add_user_message(messages, text):
    user_message = {'role': 'user', 'content': text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {'role': 'assistant', 'content': text}
    messages.append(assistant_message)

def chat(messages, system=None, stop_sequences=None):
    params = {
        "model": model_name,
        "max_tokens": 1000,
        "messages": messages
    }

    if system:
        params["system"] = system

    if stop_sequences:
        params["stop_sequences"] = stop_sequences

    with client.messages.stream(**params) as stream:
        full_response = ""

        for text in stream.text_stream:
            print(text, end="", flush=True)
            full_response += text

    return full_response
````

````python
messages = []

add_user_message(
    messages,
    "Generate a JSON schema for a user profile with name, email, and age fields."
)

add_assistant_message(messages, "```json\n")

text = chat(messages, stop_sequences=["```"])

print(text)
````
