```python
import numpy as np 
import chromadb
from sentence_transformers import SentenceTransformer
from chromadb.config import Settings
import uuid
from typing import List,Dict,Any,Tuple
from sklearn.metrics.pairwise import cosine_similarity
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

def build_chroma_vectorstore(
    docs, 
    persist_directory: str = "./chroma_db", 
    embedding_model = None
):
    """
    Chunks loaded documents and stores them in a persistent Chroma vector DB
    using modern LangChain integration packages.
    """
    if embedding_model is None:
        embedding_model = OpenAIEmbeddings()

    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200,
        length_function=len
    )
    chunks = text_splitter.split_documents(docs)
    print(f"Created {len(chunks)} text chunks from {len(docs)} document pages.")

    # langchain-chroma handles persistence automatically via persist_directory
    vectorstore = Chroma.from_documents(
        documents=chunks,
        embedding=embedding_model,
        persist_directory=persist_directory
    )
    
    print(f"Vector store successfully built and saved to '{persist_directory}'.")
    return vectorstore
```

```python
from typing import List
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

def add_documents_to_chroma(
    docs: List[Document],
    persist_directory: str = "./chroma_db",
    embedding_model = None
) -> Chroma:
    """
    Splits new documents and appends them to an existing persistent Chroma vector store.
    """
    if embedding_model is None:
        embedding_model = OpenAIEmbeddings()

    # 1. Load the existing Chroma vector store from disk
    vectorstore = Chroma(
        persist_directory=persist_directory,
        embedding_function=embedding_model
    )

    # 2. Split the new documents into chunks
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200,
        length_function=len
    )
    chunks = text_splitter.split_documents(docs)
    print(f"Created {len(chunks)} new text chunks to append.")

    # 3. Add chunks to the existing vector store
    if chunks:
        vectorstore.add_documents(documents=chunks)
        print(f"Successfully added {len(chunks)} new chunks to '{persist_directory}'.")
    else:
        print("No document chunks to add.")

    return vectorstore
```
